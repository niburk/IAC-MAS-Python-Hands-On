## 2. Decisions with `if`

Often a program should only do something under a certain **condition**. That's what `if` and `else` are for.

Two "rules":
1) After the condition comes a `:`
2) Everything that belongs to the `if` is indented (4 spaces). This is important: it doesn't just look better, it only works that way!

````python
temp = 90

if temp > 25:
    print("It's hot!")
else:
    print("It's fkn cold!")
print("this shows up anyway...")
````

### More cases with `elif`

Short for "else if". Python checks the conditions from top to bottom and takes the first condition that is true.

````python
grade = 5

if grade <= 1.5:
    print("Perfect")
elif grade <= 2.5:
    print("Good")
elif grade <= 3.5:
    print("Ok")
elif grade <= 4.0:
    print("At least you passed")
else:
    print("Failed")
````

## 3. Lists and `for` loops

### Lists
A list stores more than one value in a single variable. It is created with `[]`:

````python
shopping_list = ["Milk", "Bread", "Egg", "Cheese"]

print(shopping_list)
print(len(shopping_list))
````

You access individual elements via their position, the so-called index. !! Python starts with **0** !!

````python
print(shopping_list[0]) # 1st element
print(shopping_list[-1]) # last element

shopping_list.append("Beer") # append something
print(shopping_list)
````

````python
shopping_list.remove("Egg")
print(shopping_list)
````

### The `for` loop

The `for` loop runs the same code **for every element** in a list. Again: use `:` and indent.

````python
for article in shopping_list:
    print(f"Do not forget {article}")
````

`article` is a variable that automatically takes on the next value while running through the list. You can choose that name as you wish.

With `range()`, you can run a loop a defined number of times. Note: `range(5)` gives 0, 1, 2, 3, 4 – Python starts counting at 0 again!

````python
for i in range(5):
    print(f"Cycle No. {i}")
````

A tool often used in combination with `for` loops is summing up the elements of a list.

This works like this: `total += p` is the short form of `total = total + p`.

````python
prices = [2.49, 1.29, 3.99, 0.89]

total = 0          # don't call it "sum" – that would overwrite Python's built-in sum()!
for p in prices:
    total += p

print(f"Total: {total:.2f} €")
````

Loop + `if`: counting with conditions...

````python
temp = [18, 24, 31, 27, 15, 22, 29]  # one week

hot_days = 0
for t in temp:
    if t > 25:
        hot_days += 1

print(f"This week there were {hot_days} hot days.")
````

## 4. Functions

A **function** is like a recipe: a set of instructions for doing something that you can reuse as often as you like.

What does a function look like?

```python
def name(parameter):     # def = "define"
    ...                  # indentation
    return result        # returns a value
```

````python
def greetings(name):
    print(f"Hello {name}!")

greetings("Hans")
greetings("Megatron")
````

````python
def greetings(name, age):
    print(f"Hello {name}, who is {age} years old!")

greetings("Hans", 25)
greetings("Megatron", 13)
````

`return` returns results...

Often a function calculates something. If you want to continue working with this result, you have to tell Python to return it.

Function parameters can have a **default value**. It is used whenever the function is called without that argument.

The text in `"""` is a so-called **docstring** which is used to describe the function and what it does.

````python
def gas_volume(n, T=273.15, p=101325):
    """Volume of an ideal gas in L (n in mol, T in K, p in Pa)."""
    R = 8.314
    V = n * R * T / p     # in m³
    return V * 1000       # in L

print(f"{gas_volume(1):.1f} L")              # 0 °C, 1 atm
print(f"{gas_volume(1, T=298.15):.1f} L")    # room temperature
````

````python
import numpy as np
import matplotlib.pyplot as plt

# ============ settings ============
FILE = "sample.xy"
SKIPROWS = 0             # number of header lines in the file
WAVELENGTH = 1.5406      # Å (Cu Kα1); Mo Kα1 = 0.7093
USE_Q = True             # True: x-axis in q, False: x-axis in 2θ
NORMALIZE = True         # scale intensity to a maximum of 1
REF_PEAKS = []           # optional reference reflections in 2θ, e.g. [28.44, 47.30, 56.12]
REF_LABEL = "reference"  # legend entry for the reference reflections
OUTPUT = "xrd_pattern"   # file name without extension

# ============ style ============
plt.rcParams.update({
    "font.family": "sans-serif",
    "font.sans-serif": ["Arial", "Helvetica", "DejaVu Sans"],
    "font.size": 9,
    "axes.linewidth": 0.8,
    "xtick.direction": "in",
    "xtick.top": True,
    "xtick.major.size": 4,
    "xtick.minor.size": 2,
    "xtick.major.width": 0.8,
    "xtick.minor.width": 0.6,
    "xtick.minor.visible": True,
    "legend.frameon": False,
    "savefig.dpi": 600,
    "pdf.fonttype": 42,      # text stays editable in Inkscape/Illustrator
    "svg.fonttype": "none",
})

# ============ functions ============
def two_theta_to_q(two_theta, wavelength=1.5406):
    """Converts 2θ (in °) to q (in 1/Å)."""
    theta = np.radians(two_theta / 2)
    return 4 * np.pi * np.sin(theta) / wavelength

# ============ data ============
two_theta, intensity = np.loadtxt(FILE, unpack=True, skiprows=SKIPROWS)

if NORMALIZE:
    intensity = (intensity - intensity.min()) / (intensity.max() - intensity.min())

if USE_Q:
    x = two_theta_to_q(two_theta, WAVELENGTH)
    ref = two_theta_to_q(np.array(REF_PEAKS), WAVELENGTH)
    xlabel = r"$q$ (Å$^{-1}$)"
else:
    x = two_theta
    ref = np.array(REF_PEAKS)
    xlabel = r"2$\theta$ (°)"

# ============ plot ============
fig, ax = plt.subplots(figsize=(3.35, 2.4))

ax.plot(x, intensity, color="black", linewidth=0.8)

if len(ref) > 0:
    y_ref = -0.08 if NORMALIZE else -0.08 * intensity.max()
    ax.plot(ref, np.full_like(ref, y_ref), "|", color="tab:red",
            markersize=6, markeredgewidth=1, label=REF_LABEL)
    ax.legend(loc="upper right")

ax.set_xlim(x.min(), x.max())
ax.set_xlabel(xlabel)
ax.set_ylabel("Intensity (a.u.)")
ax.set_yticks([])        # arbitrary units → no y values

fig.tight_layout()
fig.savefig(f"{OUTPUT}.pdf")    # vector graphic for the paper
fig.savefig(f"{OUTPUT}.png")    # raster graphic for slides
plt.show()
````

## How can I apply this to my daily work?




### 1. The Ramp function

How long does my furnace need to go from T_before to T_after?

$$\text{Duration (min)} = \frac{|T_\text{after} - T_\text{before}|}{\text{rate (K/min)}}$$


````python
def ramp(T_before, T_after, rate):
    """Calculates the time (in min) needed to complete a ramp."""
    return abs(T_after - T_before) / rate
````

### 2. The program as a list

We save every segment as a **tuple**: a "list" with `()` instead of `[]` that cannot be changed.

Why? Every segment always has the same fixed structure (goal, rate, hold) and should not be changed by accident.

Bonus: each segment can be "unpacked" directly in the `for` loop: `for goal, rate, hold in program` → distributes the three values of each segment to three variables.

````python
T_START = 25     # room temperature
T_MAX = 1100     # highest possible temp of the furnace

program = [
    # (Temp goal °C, rate K/min, holding time (min))
    (700, 10, 3800),
    (25, 2, 0),
]
````

````python
time_min = 0
temp = T_START
import matplotlib.pyplot as plt

time_h = [0]      # for the plot
temps = [temp]

for goal, rate, hold_min in program:
    if goal > T_MAX:
        print(f"⚠️ {goal} °C is beyond the limit of the furnace of {T_MAX} °C!")

    # Ramp
    duration = ramp(temp, goal, rate)
    time_min += duration
    time_h.append(time_min / 60)
    temps.append(goal)

    # holding time
    if hold_min > 0:
        time_min += hold_min
        time_h.append(time_min / 60)
        temps.append(goal)

    print(f"{temp:>4} → {goal:>4} °C  |  Ramp {duration:5.0f} min  |  Hold {hold_min:4} min")
    temp = goal

print(f"\nTotal Duration: {time_min:.0f} min = {time_min / 60:.1f} h")
````

````python
from datetime import datetime, timedelta
from zoneinfo import ZoneInfo

start = datetime.now(ZoneInfo("Europe/Berlin"))
end = start + timedelta(minutes=time_min)
weekday = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]

print(f"Start:  {weekday[start.weekday()]}, {start:%d.%m. %H:%M}")
print(f"Finish: {weekday[end.weekday()]}, {end:%d.%m. %H:%M}")

````

````python
plt.figure(figsize=(8, 4))
plt.plot(time_h, temps, marker="o")
plt.axhline(T_MAX, color="red", linestyle="--", label="Max Temp.")
plt.xlabel("Time (h)")
plt.ylabel("Temp. (°C)")
plt.legend()
plt.show()
````
